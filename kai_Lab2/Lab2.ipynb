## Step 1: Load Stopwords
The `load_stopwords()` function reads the stopwords file.

```python
def load_stopwords(filename):
    stopwords = set()

    file = open(filename, "r")

    for line in file:
        word = line.strip()
        stopwords.add(word)

    file.close()

    return stopwords
```

```python
stopwords = load_stopwords("test/stopwords4T1.txt")

print("Stopwords:")
print(stopwords)
```

## Read the Text Files

The `load_words()` function reads each text file and splits the text into individual words.

```python
def load_words(filename, stopwords):
    words = []

    file = open(filename, "r")

    for line in file:
        line = line.lower()
        line = line.strip()

        line_words = line.split()

        for word in line_words:
            if word not in stopwords:
                words.append(word)

    file.close()

    return words
```

```python
words1 = load_words("test/textA4T1.txt", stopwords)
words2 = load_words("test/textB4T1.txt", stopwords)

print("Remaining words in text A:")
print(words1)

print("\nRemaining words in text B:")
print(words2)
```

## Step 2: Calculate Term Frequencies

Counts how many times each word appears in each document.

```python
def count_words(words):
    word_counts = {}

    for word in words:
        if word in word_counts:
            word_counts[word] = word_counts[word] + 1
        else:
            word_counts[word] = 1

    return word_counts
```

```python
counts1 = count_words(words1)
counts2 = count_words(words2)

print("Word counts for text A:")
print(counts1)

print("\nWord counts for text B:")
print(counts2)
```

## Term Frequency

Term frequency is the number of times a word appears divided by the total number of remaining words in the document.
For example, a word that appears 1 time in a document with 4 remaining words has a term frequency of 1/4 = 0.25.


```python
def calculate_tf(word_counts, total_words):
    tf = {}

    for word in word_counts:
        tf[word] = word_counts[word] / total_words

    return tf
```

```python
tf1 = calculate_tf(counts1, len(words1))
tf2 = calculate_tf(counts2, len(words2))

print("Term frequencies for text A:")
print(tf1)

print("\nTerm frequencies for text B:")
print(tf2)
```

## Step 3: Reduce to Unique Words

I use sets to find the words that appear in both documents.

The `&` operator finds the intersection between the two sets.


```python
common_words = set(counts1.keys()) & set(counts2.keys())

print("Words common to both documents:")
print(common_words)
```

```python
for word in common_words:
    del tf1[word]

for word in common_words:
    del tf2[word]

print("Unique words in text A:")
print(tf1)

print("\nUnique words in text B:")
print(tf2)
```
