```python
%reload_ext autoreload

from main1 import remove_stopwords, load_stopwords, get_term_freq, remove_common_words, word_freq_from_text_file #import all functions
```

### First we have a function that loads all of the stopwords into a set 

```python
stopwords = load_stopwords("stopwords.txt")
print(stopwords)
```

### Once we have that we can read the text and remove unwanted stop words

*remove_stopwords*: Reads a textFile into a string and then removes 
all the stopwords present in the given stopwords set in that string. Returns a list of words present in the textFile.

```python
words_a = remove_stopwords("textA.txt", stopwords)
words_b = remove_stopwords("textB.txt", stopwords)

print(words_a[0:5]) #Print 5 example words
```

### Now that we have the word lists for each textfile we can feed them into our next function

*get_term_freq*: Takes a list of words and returns a dictonary where the key is a word and the value is the frequency of that word. 

```python
freq_A = get_term_freq(words_a)
freq_B = get_term_freq(words_b)

print(freq_A)
```

### Finally we need to reduce the dictonaries to words that are unique and not shared between them.

*remove_common_words*: Takes in two dictonaries of words and gets the set of words that is common to each dict. Then loop through that set removing those keys from each dict. 

```python
remove_common_words(freq_A, freq_B)
```

### Our final output here than would be: 

```python
print(f"Word Frequencies A: {freq_A}")
print(f"Word Frequencies B: {freq_B}")
```
